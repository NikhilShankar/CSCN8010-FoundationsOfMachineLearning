#### **Analysing the models using bigger training, validation and test dataset.**

- Encountered an error which was occuring due to channels.
- Need to find and resolve the issue before continuing.

```python
import os
from PIL import Image
import pandas as pd

def check_image_channels(folder_path):
    # List to store image details
    image_data = []

    # Loop through all files in the folder
    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)
        
        try:
            with Image.open(file_path) as img:
                # Get the mode and size
                mode = img.mode
                channels = {
                    "1": 1,  # 1-bit pixels, black and white
                    "L": 1,  # 8-bit pixels, grayscale
                    "LA": 2,  # Grayscale + Alpha
                    "RGB": 3,  # True color
                    "RGBA": 4,  # True color with alpha
                    "CMYK": 4,  # Cyan, Magenta, Yellow, Black
                    "P": 1,  # Palette-based
                }.get(mode, "Unknown")
                
                # Append to the list
                image_data.append({
                    "File Path": file_path,
                    "Image Mode": mode,
                    "Channels": channels,
                    "Flagged": channels == 2
                })

        except Exception as e:
            # If there's an error, log it
            image_data.append({
                "File Path": file_path,
                "Image Mode": "Error",
                "Channels": "Error",
                "Flagged": False,
                "Error": str(e),
            })

    # Create a DataFrame
    df = pd.DataFrame(image_data)
    return df

# Set your folder path
folder_path = "data/RawDataset/Cat"

# Call the function
df = check_image_channels(folder_path)

# Print the DataFrame
display(df[df['Channels'] == 'Error'])
display(df['Channels'].unique())
# Set your folder path
folder_path = "data/RawDataset/Dog"

# Call the function
df = check_image_channels(folder_path)

# Print the DataFrame
display(df[df['Channels'] == 'Error'])
display(df['Channels'].unique())
```

> The error was saying that tensor encountered with an unrecognized image format with 2 channels. Hence I did the following analysis but it was not able to find any images with 2 channels which are unsupported by the preprocessor. SO shifted the start index for the dataset to exclude the initial dataset which was chosen and the error was resolved. Need to figure out what caused the issue. For now we can continue with a different section of the dataset.

#### **Conclusions**

- We trained two models in a small dataset and a large dataset.
- The custom model had 658,181 parameters which were trainable and total. The model VGG16 as base had total params of 17,992,001
out of which 5,637,121 were trainable. 
- The total parameters for VGGbase model was 27 times that of the custom model we created.
- The trainable params were ~8.5 times that of the custom model.
- The time taken for training 6000 images for custom model was ~62 minutes while VGG base model took around ~270 minutes. Time taken is almost 4.5 times that of the custom model. This was possible because  66% of the parameters in VGG model were pre-calculated.
- The accuracy was 80.67 and 98.17 for custom model and VGG model respectively. This is a phenomenal difference especially in binary classification problems. This is a solid 25% increase in accuracy. And the best point is that we achieved it using only ~4-5X increase in time. 
- The VGG-16 model trained on imagenet is a perfect choice for our problem in hand. In cases were our use case of classification is similar to the imagenet classification its a really good option to use VGG-16 trained on imagenet as a base model. 
- We can further improve accuracy precision and recall if we train it further on 12000 and eventually for the whole cats vs dogs dataset. 
- This way we make sure that the model is trained with maximum possible data available and will be powerful in predicting real world unseen data that much more.
- From analyzing the misclassified images there is an indication that the misclassification could be the result of background and other subjects presence in the picture along with the actual subject. 
- This can be reduced if we include more such images during training.
